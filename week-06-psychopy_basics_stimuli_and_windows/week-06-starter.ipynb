## 2. Hello PsychoPy

Save as `hello.py` and run with `python hello.py`

```python
# hello.py — run this as a Python script, not in Jupyter
from psychopy import visual, core, event

win = visual.Window(size=(800, 600), color='gray', units='pix')
fixation = visual.TextStim(win, text='+', color='white', height=40)
target   = visual.TextStim(win, text='Press SPACE', color='yellow', pos=(0, -50))

fixation.draw()
target.draw()
win.flip()

keys = event.waitKeys(keyList=['space', 'escape'])
print(f'Key: {keys}')

win.close()
core.quit()
```

## 3. Timing with a Clock

```python
# reaction_time.py
from psychopy import visual, core, event

win = visual.Window(size=(800, 600), color='gray', units='pix')
clock = core.Clock()

prompt = visual.TextStim(win, text='Press SPACE as fast as possible!', color='white')
prompt.draw()
win.flip()

clock.reset()
keys = event.waitKeys(keyList=['space'])
rt = clock.getTime()

result = visual.TextStim(win, text=f'RT: {rt*1000:.0f} ms', color='lime')
result.draw()
win.flip()
core.wait(2)
win.close()
core.quit()
```

## 4. Shapes and Colors

```python
# shapes.py
from psychopy import visual, core

win = visual.Window(size=(600, 600), color='black', units='pix')
rect   = visual.Rect(win,   width=200, height=100, fillColor='red',  pos=(0, 50))
circle = visual.Circle(win, radius=50,              fillColor='blue', pos=(0, -80))

rect.draw()
circle.draw()
win.flip()
core.wait(3)
win.close()
core.quit()
```
