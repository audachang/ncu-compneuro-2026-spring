## 2. Post-Run Data Analysis

```python
import pandas as pd, numpy as np

# Replace with: df = pd.read_csv('data/YOUR_PAVLOVIA_FILE.csv')
np.random.seed(42)
n = 30
df = pd.DataFrame({
    'participant': np.repeat(['P01','P02','P03','P04','P05'], n//5),
    'condition': np.tile(['congruent','incongruent'], n//2),
    'rt': np.concatenate([np.random.normal(330,45,n//2), np.random.normal(400,60,n//2)]),
    'correct': np.concatenate([np.random.binomial(1,0.94,n//2), np.random.binomial(1,0.83,n//2)])
})
print(df.groupby('condition')[['rt','correct']].agg(['mean','std']))
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

stats = df.groupby('condition')['rt'].agg(['mean','sem'])
axes[0].bar(stats.index, stats['mean'], yerr=stats['sem'], capsize=5,
            color=['steelblue','tomato'])
axes[0].set(title='Mean RT by Condition', ylabel='RT (ms)')

acc = df.groupby('condition')['correct'].mean() * 100
axes[1].bar(acc.index, acc.values, color=['steelblue','tomato'])
axes[1].set(title='Accuracy by Condition', ylabel='Accuracy (%)', ylim=(0,100))

plt.suptitle('Midterm Experiment Results')
plt.tight_layout()
plt.savefig('midterm_results.png', dpi=150)
plt.show()
```
