## 2. Cache Data for Performance

```python
# In your app.py — wrap data loading with @st.cache_data
cache_code = '''
import streamlit as st
import pandas as pd, requests

@st.cache_data(ttl=3600)   # refresh every hour
def load_data():
    r = requests.get("https://your-api-url")
    return pd.DataFrame(r.json())

df = load_data()   # fetched once, then cached
st.dataframe(df)
'''
print(cache_code)
```
