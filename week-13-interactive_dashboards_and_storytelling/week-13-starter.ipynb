## 1. Plotly Express Bar Chart

```python
import plotly.express as px
import pandas as pd, numpy as np

counties = ['Taipei','Taichung','Kaohsiung','Tainan','Hsinchu']
np.random.seed(42)
df = pd.DataFrame({'county': counties,
                   'aqi':  np.random.randint(30, 120, 5),
                   'pm25': np.random.uniform(8, 45, 5).round(1)})

fig = px.bar(df, x='county', y='aqi', color='aqi',
             color_continuous_scale='RdYlGn_r',
             title='Air Quality Index by County')
fig.update_layout(xaxis_tickangle=-30)
fig.show()
```

## 2. Scatter with Hover

```python
fig = px.scatter(df, x='pm25', y='aqi', text='county',
                 color='aqi', size='aqi',
                 color_continuous_scale='Reds',
                 title='PM2.5 vs. AQI')
fig.update_traces(textposition='top center')
fig.show()
```

## 3. Time Series with Annotation

```python
months = pd.date_range('2025-01', '2025-12', freq='MS')
np.random.seed(7)
aqi_vals = 60 + 20*np.sin(np.linspace(0, 2*3.14159, 12)) + np.random.normal(0, 5, 12)
df_t = pd.DataFrame({'month': months, 'aqi': aqi_vals})

fig = px.line(df_t, x='month', y='aqi', title='Monthly AQI Trend (Taipei)')
worst = df_t['aqi'].idxmax()
fig.add_annotation(x=df_t.loc[worst,'month'], y=df_t.loc[worst,'aqi'],
                   text='Worst month', showarrow=True, arrowhead=2)
fig.show()
```
