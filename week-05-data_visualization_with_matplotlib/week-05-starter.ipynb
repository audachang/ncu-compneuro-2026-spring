## 1. Line Plot

```python
import matplotlib.pyplot as plt
import numpy as np

blocks = np.arange(1, 9)
mean_rt = np.array([420, 405, 388, 371, 365, 360, 355, 352])

plt.figure(figsize=(7, 4))
plt.plot(blocks, mean_rt, marker='o', color='steelblue')
plt.xlabel('Block')
plt.ylabel('Mean RT (ms)')
plt.title('RT Improvement Across Blocks')
plt.tight_layout()
plt.show()
```

## 2. Histogram

```python
np.random.seed(42)
rt_con = np.random.normal(320, 40, 200)
rt_inc = np.random.normal(390, 55, 200)

fig, ax = plt.subplots(figsize=(7, 4))
ax.hist(rt_con, bins=20, alpha=0.6, label='Congruent',   color='steelblue')
ax.hist(rt_inc, bins=20, alpha=0.6, label='Incongruent', color='tomato')
ax.axvline(rt_con.mean(), color='steelblue', linestyle='--')
ax.axvline(rt_inc.mean(), color='tomato',    linestyle='--')
ax.set(xlabel='RT (ms)', ylabel='Count', title='Stroop RT Distribution')
ax.legend()
plt.tight_layout()
plt.show()
```

## 3. Bar Chart with Error Bars

```python
conditions = ['Congruent', 'Incongruent', 'Neutral']
means   = [320, 390, 355]
stderrs = [8, 12, 10]

fig, ax = plt.subplots(figsize=(5, 4))
ax.bar(conditions, means, yerr=stderrs, capsize=5,
       color=['steelblue', 'tomato', 'mediumseagreen'])
ax.set(ylabel='Mean RT (ms)', title='Stroop Effect')
plt.tight_layout()
plt.show()
```

## 4. Multi-Panel Figure

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

axes[0, 0].plot(np.arange(1,9), np.array([420,405,388,371,365,360,355,352]), 'o-')
axes[0, 0].set(title='Learning Curve', xlabel='Block', ylabel='RT (ms)')

axes[0, 1].hist(rt_con, bins=15, color='steelblue', alpha=0.7)
axes[0, 1].set(title='RT Distribution', xlabel='RT (ms)')

axes[1, 0].bar(conditions, means, color=['steelblue','tomato','mediumseagreen'])
axes[1, 0].set(title='Mean RT by Condition', ylabel='RT (ms)')

acc = [0.95, 0.87, 0.92]
axes[1, 1].scatter(means, acc, s=100, color='purple')
for i, c in enumerate(conditions):
    axes[1, 1].annotate(c, (means[i], acc[i]), xytext=(5,5), textcoords='offset points')
axes[1, 1].set(title='Speed-Accuracy', xlabel='RT (ms)', ylabel='Accuracy')

plt.suptitle('Stroop Task Summary', fontsize=14)
plt.tight_layout()
plt.savefig('stroop_summary.png', dpi=150)
plt.show()
```
