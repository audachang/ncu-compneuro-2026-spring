### Puzzle 1 — Classify a Reaction Time

Write a function `classify_rt(rt)` that returns:
- `"anticipation"` if `rt < 100` ms
- `"fast"` if `100 ≤ rt < 300` ms
- `"normal"` if `300 ≤ rt < 700` ms
- `"slow"` if `rt ≥ 700` ms

Test it on the list `[85, 210, 450, 820, 350]` and print each RT with its label.

````python
def classify_rt(rt):
    """Classify a reaction time (ms) into one of four speed categories.

    Args:
        rt (float): Reaction time in milliseconds.

    Returns:
        str: One of 'anticipation', 'fast', 'normal', or 'slow'.
    """
    if rt < 100:
        return "anticipation"
    elif rt < 300:
        return "fast"
    elif rt < 700:
        return "normal"
    else:
        return "slow"


rts = [85, 210, 450, 820, 350]
for rt in rts:
    print(f"RT={rt:4d} ms  →  {classify_rt(rt)}")
````

### Puzzle 4 — Score Responses with `zip()`

You have two lists: `stimuli` (the correct answers) and `responses` (what the participant pressed).  
Use `zip()` to iterate over both simultaneously and build a list of booleans indicating whether each response was correct.

````python
stimuli   = ["left", "right", "left", "right", "left"]
responses = ["left", "right", "right", "right", "left"]

correctness = []
for stimulus, response in zip(stimuli, responses):
    correctness.append(stimulus == response)

print("Correct responses:", correctness)
print(f"Accuracy: {sum(correctness)}/{len(correctness)} = {sum(correctness)/len(correctness)*100:.0f}%")
````

### Puzzle 7 — Skip Missing Data with `continue`

A list of RTs contains `None` to represent missed trials (no response).  
Use `continue` to skip `None` values and compute the mean only over valid entries.

````python
rts = [320, None, 415, None, 280, 510, None, 390]

total = 0
count = 0
for rt in rts:
    if rt is None:
        continue              # skip missed trials
    total += rt
    count += 1

mean = total / count if count > 0 else float("nan")
missed = len(rts) - count
print(f"Valid trials: {count}, Missed: {missed}, Mean RT: {mean:.1f} ms")
````

### Puzzle 8 — Function with Default Threshold and Docstring

Write a function `remove_outliers(rt_list, lo=100, hi=1000)` that returns a new list with RTs outside `[lo, hi]` removed.  
Include a proper docstring and demonstrate calling it with and without custom thresholds.

````python
def remove_outliers(rt_list, lo=100, hi=1000):
    """Return a filtered list with RTs outside [lo, hi] removed.

    Args:
        rt_list (list[float]): Raw reaction times in milliseconds.
        lo (float): Lower bound (inclusive). Default 100 ms.
        hi (float): Upper bound (inclusive). Default 1000 ms.

    Returns:
        list[float]: Filtered reaction times.
    """
    return [rt for rt in rt_list if lo <= rt <= hi]


raw = [85, 320, 415, 280, 1200, 510, 50, 390]

print("Default thresholds [100, 1000]:", remove_outliers(raw))
print("Custom thresholds [200, 600]: ", remove_outliers(raw, lo=200, hi=600))
````

### Puzzle 9 — Exploring Scope (LEGB Rule)

Trace through the code below **before running it** and predict what each `print()` will output.  
Then run the cell to check your predictions.  
Annotate the three scopes: Local (L), Enclosing (E), Global (G).

````python
threshold = 500   # (G) Global

def outer():
    threshold = 300   # (E) Enclosing for inner()

    def inner():
        threshold = 200   # (L) Local
        print("inner threshold:", threshold)   # What prints?

    inner()
    print("outer threshold:", threshold)        # What prints?

outer()
print("global threshold:", threshold)          # What prints?

# Expected output:
# inner threshold: 200
# outer threshold: 300
# global threshold: 500
````

### Puzzle 10 — Return Multiple Values

Write a function `rt_summary(rt_list)` that returns **four** values in one `return` statement:  
mean, standard deviation (population), minimum, and maximum.  
Use only Python built-ins (no `math` or `statistics` module) for the std calculation.

*Hint: population std = √(Σ(xᵢ − mean)² / N)*

````python
def rt_summary(rt_list):
    """Return mean, population std, min, and max of rt_list.

    Args:
        rt_list (list[float]): Reaction times in milliseconds.

    Returns:
        tuple[float, float, float, float]: (mean, std, minimum, maximum)
    """
    n = len(rt_list)
    mean = sum(rt_list) / n
    variance = sum((rt - mean) ** 2 for rt in rt_list) / n
    std = variance ** 0.5
    return mean, std, min(rt_list), max(rt_list)


rts = [320, 415, 280, 510, 390, 460, 305, 620]
mean, std, lo, hi = rt_summary(rts)
print(f"Mean: {mean:.1f} ms")
print(f"Std:  {std:.1f} ms")
print(f"Min:  {lo} ms")
print(f"Max:  {hi} ms")
````
