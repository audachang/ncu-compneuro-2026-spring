## 1. Making API Calls

```python
import requests

# Taiwan AQI open data
url = 'https://data.epa.gov.tw/api/v2/aqx_p_432'
params = {'api_key': 'anonymouskey', 'format': 'json', 'limit': '5'}

r = requests.get(url, params=params)
print('Status:', r.status_code)
if r.status_code == 200:
    data = r.json()
    print(type(data))
```

## 2. JSON to DataFrame

```python
import pandas as pd

# Simulated response — replace with real API data
sample = [
    {'site': 'Taipei',    'county': 'Taipei City',    'pm25': '12.3', 'aqi': '51'},
    {'site': 'Taichung',  'county': 'Taichung City',  'pm25': '28.7', 'aqi': '87'},
    {'site': 'Kaohsiung', 'county': 'Kaohsiung City', 'pm25': '35.1', 'aqi': '103'},
    {'site': 'Hsinchu',   'county': 'Hsinchu City',   'pm25': None,   'aqi': '-'},
]
df = pd.DataFrame(sample)
print(df)
print(df.dtypes)
```

## 3. Data Cleaning

```python
df['pm25'] = pd.to_numeric(df['pm25'], errors='coerce')
df['aqi']  = pd.to_numeric(df['aqi'],  errors='coerce')
print(df.isna().sum())
```

```python
df_clean = df.dropna(subset=['pm25'])
print(f'Rows: {len(df)} → {len(df_clean)} after dropping NaN')
print(df_clean.sort_values('pm25', ascending=False))
```
