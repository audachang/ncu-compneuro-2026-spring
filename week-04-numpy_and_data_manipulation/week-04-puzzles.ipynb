**Puzzle 5: Apply vectorized z-score normalization**

Given RTs, compute the z-score (standardize) without any loops.

Formula: `z = (x - mean) / std`

Given `rts = [320, 415, 280, 510, 390]`, print:
- Original RTs
- Mean and std
- Z-scored RTs
- Verify that z-scored mean is ~0 and std is ~1

````python
rts = np.array([320, 415, 280, 510, 390])

mean = rts.mean()
std = rts.std()

z_scores = (rts - mean) / std

print(f"Original RTs: {rts}")
print(f"Mean: {mean:.2f}, Std: {std:.2f}")
print(f"Z-scores: {z_scores}")
print(f"Z-score mean: {z_scores.mean():.6f} (should be ~0)")
print(f"Z-score std: {z_scores.std():.6f} (should be ~1)")
````

**Puzzle 6: Use boolean masking to filter valid RTs**

Given RTs including some invalid values (too fast or too slow):
```python
rts = np.array([50, 320, 415, 280, 510, 390, 1800, 100])
```

Extract only RTs in the valid range [100, 1000] ms. Print:
- Original RTs
- Valid RTs
- Number of trials removed

````python
rts = np.array([50, 320, 415, 280, 510, 390, 1800, 100])

valid_mask = (rts >= 100) & (rts <= 1000)
valid_rts = rts[valid_mask]

removed = len(rts) - len(valid_rts)

print(f"Original RTs: {rts}")
print(f"Valid RTs: {valid_rts}")
print(f"Trials removed: {removed}")
````

**Puzzle 7: Use `np.where` to label RTs as fast or slow**

Given `rts = [320, 415, 280, 510, 390]`, use `np.where` to label each RT as "fast" or "slow" relative to the mean.

Print the labels.

````python
rts = np.array([320, 415, 280, 510, 390])

mean = rts.mean()
labels = np.where(rts > mean, "slow", "fast")

print(f"RTs: {rts}")
print(f"Mean: {mean:.1f}")
print(f"Labels: {labels}")
````

**Puzzle 8: Compute statistics on a 2D array with axis parameter**

Given a 2D array of RTs (3 subjects × 4 trials):
```python
data = np.array([[320, 415, 280, 510],
                 [350, 400, 290, 480],
                 [310, 420, 300, 500]])
```

Compute:

**Puzzle 9: Use `np.random.default_rng` to simulate RT data with a seed**

Simulate 100 reaction times from a normal distribution (mean=400, std=80) using the new RNG API with seed=42.

Print:
- First 5 simulated RTs
- Actual mean and std of the simulated data
- Compare to parameters

````python
rng = np.random.default_rng(42)
rts_sim = rng.normal(loc=400, scale=80, size=100)

print(f"First 5 simulated RTs: {rts_sim[:5]}")
print(f"Simulated mean: {rts_sim.mean():.1f} (expected: 400)")
print(f"Simulated std: {rts_sim.std():.1f} (expected: 80)")
````
