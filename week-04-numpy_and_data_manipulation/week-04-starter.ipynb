## 1. Creating Arrays

```python
import numpy as np

rt = np.array([312, 289, 405, 271, 388, 450, 299])
print(rt)
print('Shape:', rt.shape)
print('Mean:', rt.mean())
print('SD:', rt.std())
```

```python
print(np.zeros(5))
print(np.ones((2, 3)))
print(np.arange(0, 1, 0.1))
print(np.linspace(0, 1, 6))
```

## 2. Indexing and Slicing

```python
rt = np.array([312, 289, 405, 271, 388, 450, 299])
print(rt[0])    # first
print(rt[-1])   # last
print(rt[2:5])  # slice

data = np.array([[312, 289, 405],
                 [271, 388, 450],
                 [299, 310, 278]])
print(data[1, :])          # row 2
print(data[:, 0])          # column 1
print(data.mean(axis=1))   # mean per row
```

## 3. Boolean Masking

```python
rt = np.array([312, 289, 405, 271, 388, 450, 299])
mask = rt < 350
print(rt[mask])               # fast trials
print(rt[rt < 350].mean())    # mean of fast trials
```

```python
mean_rt = rt.mean()
sd_rt = rt.std()
outliers = rt > mean_rt + 2 * sd_rt
print(f'Outliers: {rt[outliers]}')
print(f'N outliers: {outliers.sum()}')
```
