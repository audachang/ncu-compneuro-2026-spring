**Puzzle 3: Use string methods to clean up messy input**

Given a messy string `messy = "  CONGRUENT,  incongruent ,  NEUTRAL  "`, clean it using string methods:
- Remove leading/trailing whitespace
- Convert to lowercase
- Split by comma
- For each condition, strip whitespace and print it

Expected output:
```
congruent
incongruent
neutral
```

````python
messy = "  CONGRUENT,  incongruent ,  NEUTRAL  "

# Clean and split
conditions = messy.strip().lower().split(",")

# Print each, with whitespace removed
for condition in conditions:
    print(condition.strip())
````

**Puzzle 4: Use f-string format specifiers to format RT and accuracy**

**Puzzle 5: Perform RT arithmetic and convert units**

Given a list of RTs in milliseconds: `rts = [320, 415, 280, 510, 390]`

Compute and print:
- Mean RT in ms (1 decimal place)
- Range (max - min)
- Mean RT in seconds (3 decimal places)

Expected output:
```
Mean RT: 383.0 ms
Range: 230 ms
Mean RT in seconds: 0.383 s
```

````python
rts = [320, 415, 280, 510, 390]

mean_rt = sum(rts) / len(rts)
range_rt = max(rts) - min(rts)
mean_rt_sec = mean_rt / 1000

print(f"Mean RT: {mean_rt:.1f} ms")
print(f"Range: {range_rt} ms")
print(f"Mean RT in seconds: {mean_rt_sec:.3f} s")
````

**Puzzle 6: Demonstrate integer division and modulo for block calculations**

Given `trials = 47` (total trial count) and you want to organize them into blocks of 10:

Compute and print:
- How many complete blocks of 10?
- How many trials remain in the last incomplete block?
- For trial number 27, which block is it in? (0-indexed blocks)
- For trial 27, what position within the block? (0-indexed position)

Expected output:
```
Complete blocks: 4
Remaining trials: 7
Trial 27 is in block 2
Position within block: 7
```

````python
trials = 47
trial_number = 27

complete_blocks = trials // 10
remaining = trials % 10
block = trial_number // 10
position = trial_number % 10

print(f"Complete blocks: {complete_blocks}")
print(f"Remaining trials: {remaining}")
print(f"Trial {trial_number} is in block {block}")
print(f"Position within block: {position}")
````

**Puzzle 7: Demonstrate floating-point precision and fix with `round()`**

Show the floating-point precision issue:
- Compute `0.1 + 0.2`
- Compare it to `0.3` using `==`
- Fix it using `round()` to 1 decimal place

Expected output:
```
0.1 + 0.2 = 0.30000000000000004
Equals 0.3? False
Rounded (1 decimal): 0.3
Rounded equals 0.3? True
```

````python
result = 0.1 + 0.2

print(f"0.1 + 0.2 = {result}")
print(f"Equals 0.3? {result == 0.3}")

rounded = round(result, 1)
print(f"Rounded (1 decimal): {rounded}")
print(f"Rounded equals 0.3? {rounded == 0.3}")
````

**Puzzle 8: Type-safe conversion with validation using `try/except`**

Write code that:
- Tries to convert `"412.5"` to `float` (should succeed)
- Tries to convert `"not_a_number"` to `float` (should fail)
- Catches `ValueError` and handles it gracefully

Expected output:
```
Converted '412.5' to 412.5
Cannot convert 'not_a_number': not a valid number
```

````python
values = ["412.5", "not_a_number"]

for val in values:
    try:
        result = float(val)
        print(f"Converted '{val}' to {result}")
    except ValueError:
        print(f"Cannot convert '{val}': not a valid number")
````
