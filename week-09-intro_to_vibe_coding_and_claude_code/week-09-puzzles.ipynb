### Puzzle 1 — Create a Project Directory Tree

Write a function `scaffold_project(root)` that creates the following directory structure:

```
root/
  app.py
  requirements.txt
  README.md
  data/
    .gitkeep
  utils/
    __init__.py
    fetch_data.py
```

Use `os.makedirs()` and write placeholder content to each file.

````python
import os
import shutil

def scaffold_project(root):
    """Create a standard Streamlit project directory tree.

    Args:
        root (str): Root directory name.
    """
    os.makedirs(os.path.join(root, "data"), exist_ok=True)
    os.makedirs(os.path.join(root, "utils"), exist_ok=True)

    files = {
        "app.py":             '"""Main Streamlit application."""\nimport streamlit as st\n\nst.title("My App")\n',
        "requirements.txt":   "streamlit>=1.33\npandas>=2.0\n",
        "README.md":          "# My Project\n\nA Streamlit web app.\n",
        "data/.gitkeep":      "",
        "utils/__init__.py":  "",
        "utils/fetch_data.py": '"""Data loading utilities."""\nimport pandas as pd\n\ndef fetch_data():\n    return pd.DataFrame()\n',
    }

    for relpath, content in files.items():
        filepath = os.path.join(root, relpath)
        with open(filepath, "w", encoding="utf-8") as f:
            f.write(content)


# Test
scaffold_project("test_project")

for dirpath, dirnames, filenames in os.walk("test_project"):
    level = dirpath.replace("test_project", "").count(os.sep)
    indent = "  " * level
    print(f"{indent}{os.path.basename(dirpath)}/")
    for f in filenames:
        print(f"{indent}  {f}")

# Clean up
shutil.rmtree("test_project")
````

### Puzzle 2 — Generate a `requirements.txt` from a Package List

Write a function `write_requirements(filepath, packages)` where `packages` is a list of `(name, min_version)` tuples.

The output file should have one line per package: `name>=version`.

Test with `[("streamlit", "1.33"), ("pandas", "2.0"), ("plotly", "5.18")]`.

````python
def write_requirements(filepath, packages):
    """Write a requirements.txt file from a list of (name, version) tuples.

    Args:
        filepath (str): Output file path.
        packages (list[tuple[str, str]]): Package name and minimum version.
    """
    with open(filepath, "w", encoding="utf-8") as f:
        for name, version in packages:
            f.write(f"{name}>={version}\n")


packages = [("streamlit", "1.33"), ("pandas", "2.0"), ("plotly", "5.18")]
write_requirements("test_requirements.txt", packages)

with open("test_requirements.txt") as f:
    print(f.read())

os.remove("test_requirements.txt")
````

### Puzzle 4 — Write a `.gitignore` File Programmatically

Write a function `write_gitignore(filepath, patterns)` that creates a `.gitignore` from a list of category-pattern pairs.

Output format:
```
# Python
__pycache__/
*.pyc

# Environment
.env
.venv/
```

````python
def write_gitignore(filepath, categories):
    """Write a categorized .gitignore file.

    Args:
        filepath (str): Output file path.
        categories (list[tuple[str, list[str]]]): List of (category, patterns) tuples.
    """
    with open(filepath, "w", encoding="utf-8") as f:
        for i, (category, patterns) in enumerate(categories):
            if i > 0:
                f.write("\n")
            f.write(f"# {category}\n")
            for pattern in patterns:
                f.write(f"{pattern}\n")


categories = [
    ("Python",      ["__pycache__/", "*.pyc", "*.pyo"]),
    ("Environment", [".env", ".venv/", "venv/"]),
    ("IDE",         [".vscode/", ".idea/"]),
    ("Data",        ["data/*.csv", "data/*.json"]),
]

write_gitignore("test_gitignore", categories)

with open("test_gitignore") as f:
    print(f.read())

os.remove("test_gitignore")
````

### Puzzle 5 — Find the Bug in AI-Generated Code

The following function was "generated by an AI" to compute the mean of a list. It has **two bugs**. Identify them, explain the problems, and write a corrected version.

```python
def compute_mean(values):
    total = 0
    for v in values:
        total += v
    return total / len(values) - 1
```

Test with `[10, 20, 30]` — the correct answer is `20.0`.

````python
# Bug 1: operator precedence — total / len(values) - 1 divides first, then subtracts 1
#         Likely intended to do total / (len(values) - 1), which would be wrong for mean anyway.
# Bug 2: the formula is wrong — mean is total / len(values), not total / (len(values) - 1)
#         The "-1" should not be there at all.

# Corrected version:
def compute_mean(values):
    """Return the arithmetic mean of a list of numbers."""
    if not values:
        raise ValueError("Cannot compute mean of an empty list")
    return sum(values) / len(values)


# Test
test = [10, 20, 30]
print(f"Mean of {test} = {compute_mean(test)}")   # should be 20.0
````

### Puzzle 6 — Validate a Directory Structure

Write a function `validate_project(root)` that checks whether a project directory has all required files: `app.py`, `requirements.txt`, `README.md`, and `utils/__init__.py`.

Return a list of missing files. Test on both a valid and an incomplete scaffold.

````python
def validate_project(root):
    """Check that a project directory contains all required files.

    Args:
        root (str): Project root directory.

    Returns:
        list[str]: Paths of missing required files (empty if all present).
    """
    required = [
        "app.py",
        "requirements.txt",
        "README.md",
        os.path.join("utils", "__init__.py"),
    ]
    missing = []
    for relpath in required:
        full = os.path.join(root, relpath)
        if not os.path.isfile(full):
            missing.append(relpath)
    return missing


# Test with a complete scaffold
scaffold_project("valid_project")
missing = validate_project("valid_project")
print(f"Valid project: {len(missing)} missing files → {'✓ OK' if not missing else missing}")
shutil.rmtree("valid_project")

# Test with an incomplete scaffold
os.makedirs("incomplete_project", exist_ok=True)
with open(os.path.join("incomplete_project", "app.py"), "w") as f:
    f.write("")
missing = validate_project("incomplete_project")
print(f"Incomplete project: missing → {missing}")
shutil.rmtree("incomplete_project")
````

### Puzzle 8 — Write an Effective Prompt as a Multi-Line String

Vibe coding works best with structured, specific prompts. Write a function `format_prompt(task, inputs, outputs, constraints)` that assembles a clear AI coding prompt.

Test with a sample task: "Create a function that loads a CSV file and returns a cleaned DataFrame."

````python
def format_prompt(task, inputs, outputs, constraints):
    """Format a structured prompt for an AI coding agent.

    Args:
        task (str): What the code should do.
        inputs (list[str]): Input descriptions.
        outputs (list[str]): Output descriptions.
        constraints (list[str]): Constraints or rules.

    Returns:
        str: Formatted prompt.
    """
    lines = [f"Task: {task}", "", "Inputs:"]
    for inp in inputs:
        lines.append(f"  - {inp}")
    lines.append("")
    lines.append("Outputs:")
    for out in outputs:
        lines.append(f"  - {out}")
    lines.append("")
    lines.append("Constraints:")
    for con in constraints:
        lines.append(f"  - {con}")
    return "\n".join(lines)


prompt = format_prompt(
    task="Create a function load_and_clean(filepath) in utils/fetch_data.py",
    inputs=["filepath: str — path to a CSV file"],
    outputs=["pd.DataFrame with columns: date (datetime), station (str), pm25 (float)"],
    constraints=[
        "Use pd.to_numeric(errors='coerce') for pm25",
        "Drop rows where pm25 is NaN",
        "Do not hardcode any file paths",
    ],
)
print(prompt)
````

### Puzzle 9 — AI Code Review Checklist

Create a function `review_code(code_string)` that checks a Python code string against a basic review checklist:
1. Contains no bare `except:` (should be `except Exception:` at minimum)
2. Contains no hardcoded absolute paths (e.g., `/Users/` or `C:\\`)
3. Does not use deprecated `pandas.append()`

Return a list of issues found.

````python
def review_code(code_string):
    """Run basic code quality checks on a Python code string.

    Args:
        code_string (str): Python source code.

    Returns:
        list[str]: List of issues found (empty if clean).
    """
    issues = []

    # Check for bare except
    for i, line in enumerate(code_string.splitlines(), 1):
        stripped = line.strip()
        if stripped == "except:" or stripped.startswith("except:"):
            issues.append(f"Line {i}: bare 'except:' — use 'except Exception:' or a specific type")

    # Check for hardcoded paths
    if "/Users/" in code_string or "C:\\" in code_string:
        issues.append("Hardcoded absolute path detected — use relative paths or config")

    # Check for deprecated .append()
    if ".append(" in code_string and "pd." in code_string:
        issues.append("Possible use of deprecated pd.DataFrame.append() — use pd.concat() instead")

    return issues


# Test with problematic code
bad_code = '''import pandas as pd
df = pd.read_csv("/Users/student/data.csv")
try:
    result = df.append(new_row)
except:
    pass
'''

issues = review_code(bad_code)
print(f"Found {len(issues)} issues:")
for issue in issues:
    print(f"  ⚠ {issue}")
````

### Puzzle 10 — Generate a README from Project Metadata

Write a function `generate_readme(metadata)` where `metadata` is a dict with keys: `title`, `description`, `data_source`, `features`, `run_command`.

Return a formatted Markdown README string. Print the result.

````python
def generate_readme(metadata):
    """Generate a Markdown README from project metadata.

    Args:
        metadata (dict): Project metadata.

    Returns:
        str: Markdown-formatted README.
    """
    lines = [
        f"# {metadata['title']}",
        "",
        metadata["description"],
        "",
        "## Data Source",
        "",
        f"- {metadata['data_source']}",
        "",
        "## Features",
        "",
    ]
    for feature in metadata["features"]:
        lines.append(f"- {feature}")
    lines.extend([
        "",
        "## Run Locally",
        "",
        "```bash",
        metadata["run_command"],
        "```",
    ])
    return "\n".join(lines)


meta = {
    "title":       "Taiwan Air Quality Dashboard",
    "description": "A Streamlit app showing PM2.5 levels across Taiwan monitoring stations.",
    "data_source": "Taiwan EPA Open Data Portal (aqx_p_432)",
    "features": [
        "Filter by county and station",
        "Interactive PM2.5 trend chart",
        "AI-powered Q&A assistant",
    ],
    "run_command": "pip install -r requirements.txt && streamlit run app.py",
}

print(generate_readme(meta))
````
