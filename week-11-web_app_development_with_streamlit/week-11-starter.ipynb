## 1. Minimal App

Save as `app.py`, run with `streamlit run app.py`

```python
# app.py
import streamlit as st
import pandas as pd
import numpy as np

st.title('My First Streamlit App')

np.random.seed(42)
df = pd.DataFrame({
    'condition': ['congruent']*20 + ['incongruent']*20,
    'rt': np.concatenate([np.random.normal(320,40,20), np.random.normal(390,55,20)])
})
st.dataframe(df)
st.line_chart(df['rt'])
```

## 2. Interactive Widgets

```python
# app_interactive.py
import streamlit as st
import pandas as pd, numpy as np, matplotlib.pyplot as plt

st.title('RT Data Explorer')
condition = st.sidebar.selectbox('Condition', ['All','congruent','incongruent'])
rt_max    = st.sidebar.slider('Max RT (ms)', 200, 1000, 600)

np.random.seed(42)
df = pd.DataFrame({
    'condition': ['congruent']*20 + ['incongruent']*20,
    'rt': np.concatenate([np.random.normal(320,40,20), np.random.normal(390,55,20)])
})
if condition != 'All': df = df[df['condition'] == condition]
df = df[df['rt'] <= rt_max]

st.write(f'Showing {len(df)} trials')
st.dataframe(df)

fig, ax = plt.subplots()
ax.hist(df['rt'], bins=15, color='steelblue')
ax.set_xlabel('RT (ms)')
st.pyplot(fig)
```
