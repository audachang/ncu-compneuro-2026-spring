### Puzzle 1 — Line Plot with Mean Reference Line

Create a line plot of 50 simulated reaction times. Add a horizontal dashed line at the mean.
- Title: "RT across trials"
- X-axis: "Trial number"
- Y-axis: "Reaction time (ms)"
- Legend showing "RT" and "Mean RT"

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
trials = np.arange(1, 51)
rts = np.random.normal(loc=400, scale=60, size=50)

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(trials, rts, color="steelblue", linewidth=1.5, label="RT")
ax.axhline(rts.mean(), color="red", linestyle="--", label="Mean RT")
ax.set_xlabel("Trial number")
ax.set_ylabel("Reaction time (ms)")
ax.set_title("RT across trials")
ax.legend()
plt.tight_layout()
plt.show()
```

### Puzzle 3 — Scatter Plot with Third Variable Encoded in Color

Create a scatter plot:
- X-axis: Mean RT for 20 subjects (normal(400, 80))
- Y-axis: Accuracy for 20 subjects (uniform(0.6, 1.0))
- Color: Number of trials per subject (integers 10–60)
- Include a colorbar with label "Number of trials"
- Colormap: "viridis"

```python
rng = np.random.default_rng(42)
mean_rt  = rng.normal(400, 80, 20)
accuracy = rng.uniform(0.6, 1.0, 20)
n_trials = rng.integers(10, 60, 20)

fig, ax = plt.subplots(figsize=(8, 5))
sc = ax.scatter(mean_rt, accuracy, c=n_trials, cmap="viridis", 
                s=80, alpha=0.8, edgecolors="black", linewidth=0.5)
ax.set_xlabel("Mean RT (ms)")
ax.set_ylabel("Accuracy")
ax.set_title("Speed–Accuracy Relationship by Trial Count")
cbar = fig.colorbar(sc, ax=ax, label="Number of trials")
plt.tight_layout()
plt.show()
```

### Puzzle 6 — Multi-Panel Figure (2×2 Grid)

Create a 2×2 subplot grid with:
- **Top-left:** Line plot of RT over 50 trials
- **Top-right:** Histogram of RT distribution (20 bins)
- **Bottom-left:** Bar chart of 3 conditions
- **Bottom-right:** Scatter plot of accuracy vs. mean RT

Add a figure-level title: "Experiment Summary"

```python
np.random.seed(42)
trials = np.arange(1, 51)
rts = np.random.normal(loc=400, scale=60, size=50)
mean_rt = np.random.normal(400, 80, 20)
accuracy = np.random.uniform(0.6, 1.0, 20)
conditions = ["Congruent", "Incongruent", "Neutral"]
cond_rts = [380, 460, 400]
cond_errors = [12, 18, 10]

fig, axes = plt.subplots(2, 2, figsize=(11, 8))

# Top-left: Line plot
axes[0, 0].plot(trials, rts, color="steelblue", linewidth=1.5)
axes[0, 0].axhline(rts.mean(), color="red", linestyle="--", alpha=0.7)
axes[0, 0].set_xlabel("Trial number")
axes[0, 0].set_ylabel("RT (ms)")
axes[0, 0].set_title("RT over trials")

# Top-right: Histogram
axes[0, 1].hist(rts, bins=20, color="steelblue", edgecolor="white", alpha=0.8)
axes[0, 1].set_xlabel("RT (ms)")
axes[0, 1].set_ylabel("Count")
axes[0, 1].set_title("RT distribution")

# Bottom-left: Bar chart
axes[1, 0].bar(conditions, cond_rts, color=["#4C72B0", "#DD8452", "#55A868"],
              yerr=cond_errors, capsize=5, width=0.6, alpha=0.8)
axes[1, 0].set_ylabel("Mean RT (ms)")
axes[1, 0].set_title("Mean RT by condition")
axes[1, 0].set_ylim(0, 550)

# Bottom-right: Scatter plot
axes[1, 1].scatter(mean_rt, accuracy, color="coral", edgecolors="black", s=60, alpha=0.8)
axes[1, 1].set_xlabel("Mean RT (ms)")
axes[1, 1].set_ylabel("Accuracy")
axes[1, 1].set_title("Speed–accuracy relationship")

fig.suptitle("Experiment Summary", fontsize=14, fontweight="bold", y=1.00)
plt.tight_layout()
plt.show()
```

### Puzzle 7 — Annotate with Text and Arrow

Create a line plot of 50 RTs:
- Find the slowest trial (highest RT)
- Annotate it with text "Slowest trial" and an arrow pointing to it
- Use `annotate()` with `arrowprops`

```python
np.random.seed(42)
trials = np.arange(1, 51)
rts = np.random.normal(loc=400, scale=60, size=50)

max_idx = np.argmax(rts)
max_trial = trials[max_idx]
max_rt = rts[max_idx]

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(trials, rts, color="steelblue", linewidth=1.5, marker="o", markersize=4)
ax.annotate(
    "Slowest trial",
    xy=(max_trial, max_rt),
    xytext=(max_trial - 10, max_rt + 40),
    arrowprops=dict(arrowstyle="->", color="red", lw=1.5),
    fontsize=10,
    color="red",
    fontweight="bold"
)
ax.set_xlabel("Trial number")
ax.set_ylabel("RT (ms)")
ax.set_title("RT over trials with annotation")
plt.tight_layout()
plt.show()
```

### Puzzle 8 — Shared Y-Axis

Create two histograms side by side:
- Left: Congruent RTs (normal(380, 60, 100))
- Right: Incongruent RTs (normal(460, 80, 100))
- Use `sharey=True` to link the y-axes for fair comparison
- Add titles and labels to each

```python
np.random.seed(42)
rts_congruent   = np.random.normal(380, 60, 100)
rts_incongruent = np.random.normal(460, 80, 100)

fig, axes = plt.subplots(1, 2, figsize=(10, 4), sharey=True)

axes[0].hist(rts_congruent, bins=20, color="steelblue", edgecolor="white", alpha=0.8)
axes[0].set_xlabel("RT (ms)")
axes[0].set_ylabel("Count")
axes[0].set_title("Congruent condition")

axes[1].hist(rts_incongruent, bins=20, color="coral", edgecolor="white", alpha=0.8)
axes[1].set_xlabel("RT (ms)")
axes[1].set_title("Incongruent condition")

fig.suptitle("RT Distributions", fontsize=12, fontweight="bold")
plt.tight_layout()
plt.show()
```

### Puzzle 9 — Save Figure to Multiple Formats

Create a simple scatter plot and save it:
- Save to PNG with dpi=150
- Save to PDF with default dpi
- Check that both files exist and compare file sizes
- Use `bbox_inches="tight"` for both

```python
import os

np.random.seed(42)
mean_rt = np.random.normal(400, 80, 20)
accuracy = np.random.uniform(0.6, 1.0, 20)

fig, ax = plt.subplots(figsize=(6, 4))
ax.scatter(mean_rt, accuracy, color="coral", edgecolors="black", s=100, alpha=0.8)
ax.set_xlabel("Mean RT (ms)")
ax.set_ylabel("Accuracy")
ax.set_title("Speed–Accuracy Relationship")

# Save to both formats
png_file = "experiment_summary.png"
pdf_file = "experiment_summary.pdf"

fig.savefig(png_file, dpi=150, bbox_inches="tight")
fig.savefig(pdf_file, bbox_inches="tight")

# Check file sizes
png_size = os.path.getsize(png_file)
pdf_size = os.path.getsize(pdf_file)

print(f"PNG size: {png_size:,} bytes")
print(f"PDF size: {pdf_size:,} bytes")
print(f"Ratio (PNG/PDF): {png_size/pdf_size:.2f}")

plt.show()
```
