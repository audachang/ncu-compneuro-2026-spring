### 情境

```python
spike_train = [
    0, 0, 1, 1, 1, 0, 0, 0, 0, 1,   # 0-100 ms
    1, 0, 0, 0, 0, 0, 0, 1, 1, 1,   # 100-200 ms
```

#### (a) 計算 Firing Rate
- Firing rate = spike 總數 ÷ 總時間（秒）
- 總時間 = 資料點數量 × 10 ms
- 以 `Hz`（spikes/sec）為單位輸出

- **Burst** 定義：**連續 3 個（含）以上** 的 `1`
- 找出每個 burst 的「**起始 index**」和「**長度**」
- 例如：如果 index 2, 3, 4 都是 1，那就是一個起始於 index 2、長度為 3 的 burst

#### (c) 找出最長的 Inter-Spike Interval (ISI)
- **ISI** 定義：兩個 spike 之間連續 `0` 的數量
- 找出最長的 ISI 是多少個時間點（× 10 ms 換算為毫秒）
- 也印出這段最長 ISI 的起始 index

#### (d) 分類神經元
根據 firing rate 分類：
- firing rate < 10 Hz → `"Low-firing neuron"`
- 10 Hz ≤ firing rate < 30 Hz → `"Medium-firing neuron"`  
- firing rate ≥ 30 Hz → `"High-firing neuron"`
