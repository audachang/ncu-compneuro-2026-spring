## 2. Full Stroop Task Template

Save as `stroop.py` and run with `python stroop.py`

```python
# stroop.py
from psychopy import visual, core, event
import pandas as pd, random

win = visual.Window(size=(800, 600), color='gray', units='pix')
clock = core.Clock()
fixation  = visual.TextStim(win, text='+',  color='white', height=40)
word_stim = visual.TextStim(win, text='',   height=60)

trials = [
    {'word': 'RED',  'color': 'red',  'condition': 'congruent'},
    {'word': 'BLUE', 'color': 'blue', 'condition': 'congruent'},
    {'word': 'RED',  'color': 'blue', 'condition': 'incongruent'},
    {'word': 'BLUE', 'color': 'red',  'condition': 'incongruent'},
] * 5
random.shuffle(trials)

results = []
for t in trials:
    fixation.draw(); win.flip(); core.wait(0.5)
    word_stim.text = t['word']
    word_stim.color = t['color']
    word_stim.draw(); win.flip()
    clock.reset()
    keys = event.waitKeys(maxWait=2.0, keyList=['r','b','escape'], timeStamped=clock)
    if not keys or keys[0][0] == 'escape': break
    key, rt = keys[0]
    t['key'] = key
    t['rt']  = round(rt * 1000, 1)
    t['correct'] = (key == ('r' if t['color']=='red' else 'b'))
    results.append(t)

pd.DataFrame(results).to_csv('stroop_results.csv', index=False)
win.close(); core.quit()
```
