## 2. Writing Tests with pytest

```python
# data_utils.py
def filter_by_condition(df, col, value):
    return df[df[col] == value].copy()

def compute_mean(df, col):
    return df[col].mean()
```

```python
# test_data_utils.py
import pandas as pd

def test_filter_by_condition():
    df = pd.DataFrame({'condition': ['A','B','A'], 'rt': [300,400,310]})
    result = filter_by_condition(df, 'condition', 'A')
    assert len(result) == 2
    assert all(result['condition'] == 'A')

def test_compute_mean():
    df = pd.DataFrame({'rt': [300, 400, None]})
    assert compute_mean(df, 'rt') == 350.0

# Run with: pytest -v
```

## 3. GitHub Actions Workflow

```python
# Save this to .github/workflows/test.yml
workflow = '''
name: Run Tests
on: [push, pull_request]
jobs:
  test:
    runs-on: ubuntu-latest
    steps:
      - uses: actions/checkout@v4
      - uses: actions/setup-python@v4
        with:
          python-version: "3.10"
      - run: pip install -r requirements.txt
      - run: pytest tests/ -v
'''
print(workflow)
```
