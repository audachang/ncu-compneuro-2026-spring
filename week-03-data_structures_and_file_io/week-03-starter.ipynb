## 3. File I/O

```python
with open('notes.txt', 'w') as f:
    f.write('Participant: P01\n')
    f.write('Date: 2026-02-01\n')

with open('notes.txt', 'r') as f:
    print(f.read())
```

```python
import csv
with open('results.csv', 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(['participant', 'rt', 'correct'])
    writer.writerow(['P01', 312, True])

with open('results.csv', 'r') as f:
    for row in csv.DictReader(f):
        print(row)
```

## 4. Tuples and Sets

```python
point = (320, 240)
x, y = point
print(f'x={x}, y={y}')
```

```python
all_conds = ['congruent', 'incongruent', 'congruent', 'neutral']
print(set(all_conds))
A = {'P01', 'P02', 'P03'}
B = {'P02', 'P03', 'P04'}
print(A & B)  # intersection
print(A | B)  # union
```
