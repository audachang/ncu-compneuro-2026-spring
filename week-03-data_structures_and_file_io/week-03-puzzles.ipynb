### Puzzle 1 — List Statistics

Given a list of reaction times, compute and print: length, mean, min, max, and the range (max − min).  
Then produce a sorted copy (ascending) **without modifying the original list**.

````python
rts = [320, 415, 280, 510, 390, 460, 305, 620]

n      = len(rts)
mean   = sum(rts) / n
lo, hi = min(rts), max(rts)

print(f"Count : {n}")
print(f"Mean  : {mean:.1f} ms")
print(f"Min   : {lo} ms")
print(f"Max   : {hi} ms")
print(f"Range : {hi - lo} ms")

sorted_rts = sorted(rts)           # sorted() returns a new list
print(f"\nOriginal: {rts}")
print(f"Sorted  : {sorted_rts}")
````

### Puzzle 4 — Accuracy and Mean RT from a List of Dicts

Compute per-condition accuracy and mean RT from a list-of-dicts dataset using list comprehensions and `dict.get()`.

````python
results = [
    {"condition": "congruent",   "rt": 310, "correct": True},
    {"condition": "incongruent", "rt": 520, "correct": False},
    {"condition": "congruent",   "rt": 340, "correct": True},
    {"condition": "incongruent", "rt": 490, "correct": True},
    {"condition": "congruent",   "rt": 290, "correct": True},
    {"condition": "incongruent", "rt": 610, "correct": False},
]

conditions = ["congruent", "incongruent"]

for cond in conditions:
    subset  = [r for r in results if r["condition"] == cond]
    n_corr  = sum(r["correct"] for r in subset)
    acc     = n_corr / len(subset) * 100
    mean_rt = sum(r["rt"] for r in subset) / len(subset)
    print(f"{cond:12s}  accuracy={acc:.0f}%  mean RT={mean_rt:.1f} ms")
````

### Puzzle 5 — List Comprehension: Filter, Map, and Combined

Using list comprehensions:
1. Filter out RTs outside 100–1000 ms
2. Convert valid RTs from ms to seconds
3. Produce a combined filtered-and-converted list in one expression

````python
raw = [85, 320, -10, 415, 1200, 280, 510, None, 390]

# 1. Filter: only numeric values in [100, 1000]
valid = [rt for rt in raw if isinstance(rt, (int, float)) and 100 <= rt <= 1000]
print("Valid RTs (ms):", valid)

# 2. Map: convert to seconds
valid_sec = [rt / 1000 for rt in valid]
print("Valid RTs (s): ", [round(s, 3) for s in valid_sec])

# 3. Combined in one comprehension
valid_sec_combined = [
    rt / 1000
    for rt in raw
    if isinstance(rt, (int, float)) and 100 <= rt <= 1000
]
print("Combined (s):  ", [round(s, 3) for s in valid_sec_combined])
````

### Puzzle 6 — Grouping Trials with `defaultdict`

Use `collections.defaultdict` to group all RTs by condition.  
Then print each condition's trial count and mean RT.

````python
from collections import defaultdict

results = [
    {"condition": "congruent",   "rt": 310},
    {"condition": "incongruent", "rt": 520},
    {"condition": "neutral",     "rt": 380},
    {"condition": "congruent",   "rt": 295},
    {"condition": "incongruent", "rt": 490},
    {"condition": "neutral",     "rt": 340},
    {"condition": "congruent",   "rt": 330},
    {"condition": "incongruent", "rt": 610},
]

by_condition = defaultdict(list)
for r in results:
    by_condition[r["condition"]].append(r["rt"])

for cond, rts in sorted(by_condition.items()):
    mean_rt = sum(rts) / len(rts)
    print(f"{cond:12s}  n={len(rts)}  mean RT={mean_rt:.1f} ms")
````

### Puzzle 7 — Tuple Packing, Unpacking, and as a Dict Key

Demonstrate:  
1. Packing and unpacking a trial summary tuple  
2. Using a tuple `(subject, condition)` as a dictionary key  
3. Looking up a value with `.get()` to handle missing combinations safely

````python
# 1. Packing and unpacking
summary = ("P01", "congruent", 342.1, 0.93)   # (subject, condition, mean_rt, accuracy)
subject, condition, mean_rt, accuracy = summary
print(f"Subject: {subject}, Condition: {condition}, Mean RT: {mean_rt} ms, Accuracy: {accuracy:.0%}")

# 2. Tuple as dict key
lookup = {
    ("P01", "congruent"):   [310, 295, 330],
    ("P01", "incongruent"): [510, 490, 610],
    ("P02", "congruent"):   [280, 300, 315],
}

key = ("P01", "congruent")
rts = lookup[key]
print(f"\n{key} → RTs: {rts}, mean: {sum(rts)/len(rts):.1f} ms")

# 3. Safe lookup for missing combination
missing_key = ("P03", "neutral")
rts = lookup.get(missing_key, [])   # returns [] instead of raising KeyError
print(f"{missing_key} → {rts if rts else 'No data'}")
````

### Puzzle 8 — Set Operations: Participant Tracking

Use set operations to answer three questions about a multi-session study:  
1. Which participants completed **both** sessions?  
2. Which participants **only** completed session 1?  
3. Which participants completed **at least one** session?

````python
session1 = {"P01", "P02", "P03", "P04", "P05"}
session2 = {"P02", "P03", "P05", "P06"}

both         = session1 & session2          # intersection
only_session1 = session1 - session2         # difference
either       = session1 | session2          # union
only_session2 = session2 - session1         # new in session 2

print(f"Completed both sessions   : {sorted(both)}")
print(f"Only session 1 (dropped)  : {sorted(only_session1)}")
print(f"New in session 2          : {sorted(only_session2)}")
print(f"At least one session      : {sorted(either)}")
print(f"Session 2 ⊆ Session 1?    : {session2.issubset(session1)}")
````

### Puzzle 9 — Write and Read a Text Log File

Write a function `log_trial(filepath, trial_num, rt, correct)` that **appends** one trial record to a text file.  
Then simulate 5 trials, log each one, and read the file back to verify.

````python
import os

LOG_FILE = "trial_log.txt"

def log_trial(filepath, trial_num, rt, correct):
    """Append a single trial record to a text file.

    Args:
        filepath (str): Path to the log file.
        trial_num (int): 1-indexed trial number.
        rt (float): Reaction time in ms.
        correct (bool): Whether the response was correct.
    """
    with open(filepath, "a", encoding="utf-8") as f:
        f.write(f"Trial {trial_num:03d}  RT={rt:6.1f} ms  correct={correct}\n")


# Remove any existing log from a previous run
if os.path.exists(LOG_FILE):
    os.remove(LOG_FILE)

# Simulate 5 trials
fake_data = [(320, True), (510, False), (390, True), (280, True), (750, False)]
for i, (rt, correct) in enumerate(fake_data, start=1):
    log_trial(LOG_FILE, i, rt, correct)

# Read back and verify
print("Contents of", LOG_FILE)
print("-" * 40)
with open(LOG_FILE, encoding="utf-8") as f:
    for line in f:
        print(line, end="")
````

### Puzzle 10 — Write and Read a CSV File with `DictWriter` / `DictReader`

1. Write a list-of-dicts dataset to `results.csv` using `csv.DictWriter` (remember `newline=""`)  
2. Read it back with `csv.DictReader`, converting each value to the correct Python type  
3. Verify the round-trip by comparing the original and reloaded data

````python
import csv

CSV_FILE = "results.csv"

original = [
    {"trial": 1, "condition": "congruent",   "rt_ms": 310.0, "correct": True},
    {"trial": 2, "condition": "incongruent", "rt_ms": 520.5, "correct": False},
    {"trial": 3, "condition": "congruent",   "rt_ms": 295.0, "correct": True},
    {"trial": 4, "condition": "neutral",     "rt_ms": 380.2, "correct": True},
]

fieldnames = ["trial", "condition", "rt_ms", "correct"]

# --- Write ---
with open(CSV_FILE, "w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(original)

print(f"Wrote {len(original)} rows to {CSV_FILE}\n")

# --- Read back with type conversion ---
loaded = []
with open(CSV_FILE, encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        loaded.append({
            "trial":     int(row["trial"]),
            "condition": row["condition"],
            "rt_ms":     float(row["rt_ms"]),
            "correct":   row["correct"] == "True",
        })

# --- Verify ---
print("Round-trip check:")
for orig, load in zip(original, loaded):
    match = orig == load
    print(f"  Trial {orig['trial']}: {'✓ match' if match else '✗ MISMATCH'}")
````
